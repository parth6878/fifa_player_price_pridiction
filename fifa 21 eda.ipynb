#### converting height and weight column to int

```python
df.Weight.unique()
```

```python
df.Height.unique()
```

```python
def clean_height(x):
    try:
        return int((int(x[0]) * 30.48) + (int(x[1]) * 2.54))
    except:
        return int(x[0])
```

```python
df["Height"]=df["Height"].apply(lambda x: x.split('c')[0])
df["Height"]=df["Height"].apply(lambda x: x.rstrip("\"")).apply(lambda x: x.split("'")).apply(clean_height)
df["Weight"]=df["Weight"].apply(lambda x: x.split('k')[0])
df["Weight"]=df["Weight"].apply(lambda x: x.split('l')[0]).apply(lambda x: int(x))
```

```python
df.Value.unique()
```

```python
def changek(x):
    try:
        if "M" in x:
            s=x.rstrip("M")
            return float(s) * 1000000.00
        elif "K" in x:
            s=x.rstrip("K")
            return float(s) *1000.00
        else:
            return float(x)
    except:
        return int(x)
```

```python
df.Value=df.Value.apply(lambda x: x.lstrip("€")).apply(changek)
df.Wage=df.Wage.apply(lambda x: x.lstrip("€")).apply(changek)
df["Release Clause"]=df["Release Clause"].apply(lambda x: x.lstrip("€")).apply(changek)
```

```python
df.Hits=df.Hits.fillna(0)
```

```python
df.Hits=df.Hits.apply(changek)
```

#### changing joined to date time

```python
df.Joined.unique()
```

```python
def clean_datetime(x):
    months={"Jan":1,"Feb":2,"Mar":3,"Apr":4,"May":5,"Jun":6,"Jul":7,"Aug":8,"Sep":9,"Oct":10,"Nov":11,"Dec":12}
    year=x.split(" ")[-1]
    date=x.split(" ")[1].rstrip(",")
    month=x.split()[0]
    for i in months:
        if i==month:
            month=months[i]
    return (f"{date},{month},{year}")
```

```python
df.Joined=df.Joined.apply(clean_datetime).apply(lambda x: x.replace(",","-"))
```

```python
df.Joined=pd.to_datetime(df.Joined, format='%d-%m-%Y')
```

# model building

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
```

```python

candidate_features = ["OVA","POT","Age","Attacking","Skill","Movement","Power",
                       "Mentality","Defending","Goalkeeping","Height","Weight",
                       "W/F","SM","IR"]

X = df[candidate_features]
y = df["Value_log"]

rf = RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X, y)

importances = pd.Series(rf.feature_importances_, index=candidate_features).sort_values(ascending=False)

plt.figure(figsize=(8, 6))
sns.barplot(x=importances.values, y=importances.index)
plt.title("Feature importance for predicting log(Value) — diagnostic Random Forest")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

importances
```
